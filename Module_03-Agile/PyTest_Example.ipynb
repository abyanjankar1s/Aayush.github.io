## The Story

"As a user, I want to calculate the total runtime of a road trip playlist from a collection of song durations, and be warned if any single track takes up more than half of the total listening time."

## The Tasks

Task 1: Calculate total runtime What it does: Add up all song durations to get the playlist's total time. Correct looks like: Given a list of durations, returns their sum.

Task 2: Find the longest track's share What it does: Determine what fraction of the total one track represents. Correct looks like: Given a list of durations, returns the largest single duration divided by the total.

Task 3: Warn if imbalanced What it does: Flags whether the longest track exceeds half the total listening time. Correct looks like: Returns True/False based on whether Task 2's result is greater than 0.5.

Notice Task 3 depends on Task 2, and Task 2 depends on Task 1 — that's normal. Tasks can build on each other; they just need to be separately testable.

```python
def test_calculate_total_runtime():
    # Arrange 1: Normal case
    durations_normal = [10, 20, 30]
    expected_total_normal = 60

    # Act 1
    actual_total_normal = calculate_total_runtime(durations_normal)

    # Assert 1
    assert actual_total_normal == expected_total_normal

    # Arrange 2: Empty list case
    durations_empty = []
    expected_total_empty = 0

    # Act 2
    actual_total_empty = calculate_total_runtime(durations_empty)

    # Assert 2
    assert actual_total_empty == expected_total_empty

def test_find_longest_track_share():
    # Arrange
    durations = [10, 20, 30]
    # Assuming calculate_total_runtime exists and works correctly
    total_runtime = 60 # From Task 1: calculate_total_runtime(durations)
    expected_share = 30 / 60  # Longest track (30) / total (60)

    # Act
    actual_share = find_longest_track_share(durations, total_runtime)

    # Assert
    assert actual_share == expected_share

def test_warn_if_imbalanced():
    # Arrange
    longest_track_share_high = 0.6  # Greater than 0.5
    longest_track_share_low = 0.4   # Less than 0.5

    # Act
    warning_high = warn_if_imbalanced(longest_track_share_high)
    warning_low = warn_if_imbalanced(longest_track_share_low)

    # Assert
    assert warning_high is True
    assert warning_low is False
```

```python
def test_calculate_total_runtime_invalid_input_type():
    # Arrange
    invalid_durations = [10, '20', 30] # List containing a non-integer

    # Act & Assert
    # We expect a TypeError because '20' cannot be summed with integers
    try:
        calculate_total_runtime(invalid_durations)
        assert False, "TypeError was not raised for invalid input type"
    except TypeError:
        assert True

    # Arrange 2: Non-list input
    non_list_input = "not a list"

    # Act & Assert 2
    # We expect a TypeError because sum() cannot iterate over a string
    try:
        calculate_total_runtime(non_list_input)
        assert False, "TypeError was not raised for non-list input"
    except TypeError:
        assert True
```

```python
def calculate_total_runtime(durations: list[int]) -> int:
    """Adds up all song durations to get the playlist's total time."""
    return sum(durations)

def find_longest_track_share(durations: list[int], total_runtime: float) -> float:
    """Determines what fraction of the total one track represents.

    Assumes total_runtime is already calculated from the durations.
    """
    if not durations or total_runtime == 0:
        return 0.0
    longest_duration = max(durations)
    return longest_duration / total_runtime

def warn_if_imbalanced(longest_track_share: float) -> bool:
    """Flags whether the longest track exceeds half the total listening time."""
    return longest_track_share > 0.5
```
